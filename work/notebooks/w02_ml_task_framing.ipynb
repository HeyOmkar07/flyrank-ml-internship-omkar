## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*



**Task type: Ranking / Scoring**

The ML task is to rank website content pages by their likelihood of being
underperforming or requiring optimization attention. The model would assign
each page an opportunity score, allowing the content/SEO team to prioritize
which pages should be reviewed first.

This is better framed as a ranking/scoring problem because the business does
not only need a yes/no prediction. It needs an ordered list of pages so that
limited content-team resources can be focused on the highest-priority pages.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

**Ideal target:**

Whether a page is genuinely underperforming in terms of clicks relative to
its expected potential, given its search position and other contextual factors.

**Measurable proxy:**

The primary proxy is `ctr_gap`, representing the difference between a page's
observed CTR and the expected CTR for comparable pages in the same position
tier.

A strongly negative `ctr_gap` indicates that a page is receiving fewer clicks
than comparable pages at similar positions.

**Proxy limitations:**

- Low CTR may be caused by SERP features or competing results rather than poor
  content.
- Search intent mismatch can reduce CTR even when the page itself is good.
- Position tier may not fully capture all factors affecting expected CTR.
- Therefore, `ctr_gap` should be treated as a useful proxy rather than a
  perfect ground-truth measure of content quality.

## 3. Success Metric

**Primary metric: Precision@K**

I would evaluate the ranking using Precision@K, particularly Precision@20
and Precision@50.

Precision@K measures the fraction of the top K pages selected by the model
that are actually relevant according to the available evaluation label.

For example, Precision@20 answers:

"Of the 20 pages the model ranked as the highest priority, how many were
actually identified as relevant?"

This metric is appropriate because the content team has limited resources and
will review only a relatively small number of pages first.

For the current starter dataset, the available `is_declining_label` can be
used as a bootstrap evaluation label. This evaluates decline detection rather
than proving that the model has perfectly identified the true CTR opportunity.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule such as "prioritize every page with CTR below 2%" is easy to
understand but does not account for differences in search position, traffic
volume, trends, and other page-level signals.

For example, a CTR of 2% may be poor for a page ranking in position 2 but
reasonable for a page ranking much lower in the search results.

An ML-based scoring approach can combine multiple signals and learn patterns
from historical examples rather than relying on a single manually selected
threshold.

The main advantage is therefore not that ML is always better than rules, but
that ML can provide a more flexible prioritization system when multiple
signals interact and the relationship between those signals and the desired
outcome is difficult to capture with one fixed threshold.

A fixed rule remains useful as a baseline that the ML approach should be
required to outperform.

IF CTR < 2%
THEN optimize page
